```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
# we want in this format - ['Action','Adventure','Fantasy','Science Fiction']
import ast
def extract_name(obj):
    l = []
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
extract_name('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
movies['genres'] = movies['genres'].apply(extract_name)
```

```python
def extract_top_3_names(obj):
    l=[]
    count=0
    for i in ast.literal_eval(obj):
        l.append(i['name'])
        count+=1
        if(count == 3):
            break
    return l
```

```python
movies['cast'] = movies['cast'].apply(extract_top_3_names)
```

```python
def extract_director_name(obj):
    l = []
    for i in ast.literal_eval(obj):
        if(i['job'] == 'Director'):
            l.append(i['name'])
            break
    return l
```

```python
movies['crew'] = movies['crew'].apply(extract_director_name)
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
stem('a cryptic message from bond’s past sends him on a trail to uncover a sinister organization. while m battles political forces to keep the secret service alive, bond peels back the layers of deceit to reveal the terrible truth behind spectre. action adventure crime spy basedonnovel secretagent sequel mi6 britishsecretservice unitedkingdom danielcraig christophwaltz léaseydoux sammendes'
)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow = CountVectorizer(max_features = 5000,stop_words='english')
```

```python
vectors_bow = bow.fit_transform(new_df['tags']).toarray()
```

```python
vectors_bow
```

```python
vectors_bow[0]
```

```python
bow.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
bow_similarity = cosine_similarity(vectors_bow)
```

```python
bow_similarity.shape
```

```python
bow_similarity[0]
```

```python
bow_similarity[1]
```

```python
def recommend(movie,top_n):
    matches = new_df[new_df['title'].str.lower() == movie.lower()]

    if len(matches) == 0:
        print("Movie not found!")
        return
    
    # Get the index of the movie
    movie_index = matches.index[0]
    distances = bow_similarity[movie_index]
    
    # Get top_n recommendations (skipping the first one as it is the movie itself)
    movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:top_n+1]
    
    for i in movie_list:
        # Correctly access the title using iloc[index]
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar',10)
```
